```python
# Load the CSV file into a DataFrame
df = pd.read_csv('ces_2002-2013_monthly_2025421.csv') 

# SUMMING UP FARM EMPLOYMENT IN 2013 -----------------------------------------

# sum up farm employment per county per year
top_6_counties = ['Kern County', 'Los Angeles County', 'Riverside County', 'San Diego County', 'San Joaquin County',
                  'Tulare County']
percent_farm = []
for county in top_6_counties :
    farmcounty2013 = df.loc[
        (df['Area Name'] == county) & (df['Year'] == 2013) & (df['Industry Title'] == 'Total Farm'), 
        ['Month','Current Employment']]
    farmcounty2013 = pd.DataFrame(farmcounty2013)
    percent_farm.append(farmcounty2013)
    #print("TOTAL NUMBER OF EMPLOYMENT IN FARM WORK IN", county, "IN 2013: ", farmcounty2013)

# SUMMING UP FARM EMPLOYMENT IN 2023 ------------------------------------------

# Load the CSV file into a DataFrame
df = pd.read_csv('ces_2014-2025_monthly_2026123.csv') 

# sum up farm employment per county per year
top_6_counties = ['Kern County', 'Los Angeles County', 'Riverside County', 'San Diego County', 'San Joaquin County',
                  'Tulare County']
percent_farm2 = []
for county in top_6_counties :
    farmcounty2023 = df.loc[
        (df['Area Name'] == county) & (df['Year'] == 2023) & (df['Industry Title'] == 'Total Farm'), 
        ['Month','Current Employment']]
    farmcounty2023 = pd.DataFrame(farmcounty2023)
    percent_farm2.append(farmcounty2023)
   # print("TOTAL NUMBER OF EMPLOYMENT IN FARM WORK IN", county, "IN 2023: /n", farmcounty2023)
#percent_farm
```
